# Токенизация в NLP

**Токенизация** – процесс разбиения текста на минимальные единицы (токены): слова, подслова или символы


## 1. Простые методы токенизации

### 1.1. Разделение по пробелам


```python
text = "Hello, world! This is a test."
text.split()
```


### 1.2. Регулярные выражения


```python
import re

text = "Hello, world! This is a test."
tokens = re.findall(r"\w+", text)
tokens
```


**Ограничения простых методов**:  
- Не учитывают пунктуацию
- Не работают с разными языками и морфологией
- Не подходят для серьёзных задач

- **Применение**: образовательные проекты, простая предобработка текста
- **Ограничения**: работает медленно на больших корпусах

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize

text = "Hello, world! This is a test."
print(word_tokenize(text))
print(sent_tokenize(text))
```

## 5. Пример предобработки текста

Возьмём новостной текст и применим полный пайплайн предобработки: очистка → токенизация → лемматизация → удаление стоп-слов


```python
import spacy
from nltk.corpus import stopwords
import nltk
import re

nltk.download('stopwords')

# Загрузка модели spaCy для английского
nlp = spacy.load("en_core_web_sm")
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # 1. Приведение к нижнему регистру
    text = text.lower()
    # 2. Удаление пунктуации и цифр
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # 3. Токенизация и лемматизация через spaCy
    doc = nlp(text)
    # 4. Удаление стоп-слов и коротких токенов
    tokens = [token.lemma_ for token in doc if token.text not in stop_words and len(token.text) > 2]
    return tokens

# Пример
sample_text = "The quick brown fox jumps over the lazy dog. This is a simple sentence for preprocessing!"
processed = preprocess_text(sample_text)
print("Оригинал:", sample_text)
print("После предобработки:", processed)
```

1. Out-of-Vocabulary - те слова языка, которые не входят в список самых частотных слов языка, исользуемых для обучения переводческой модели, описанной в сатье, и которые переводятся не при помощи словарного соответсвия в словаре другого языка, а при помощи посимвольного парсинга и анализа слова.
2. Авторы исследования предложили создать гибридную словесно-символьную переводческую модель, которая будет иметь небольшой словарь самых частотных слов обоих языков, а все более редкие слова она будет переводить при помощи парсинга каждого символа слова и учета вектора контекста и последовательности символов. Такой подход позволил добиться улучшения метрик оценки машинного перевода и показал, что он справляется с ним лучше, чем просто словесные модели.
